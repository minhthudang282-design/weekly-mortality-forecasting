## 3.1 Setup — Load dữ liệu vào SQLite

```python
import pandas as pd
import sqlite3

# All SQL descriptive counts use observed registrations, never interpolated rows.
df = pd.read_csv('uk_weather_health_weekly_2016_2024_TOTAL_FINAL.csv',
                 parse_dates=['week_start_sat','week_start_mon','week_end'])
df['year'] = df['week_start_mon'].dt.year
df['month'] = df['week_start_mon'].dt.month
season_map = {12:'Winter',1:'Winter',2:'Winter',3:'Spring',4:'Spring',5:'Spring',
              6:'Summer',7:'Summer',8:'Summer',9:'Autumn',10:'Autumn',11:'Autumn'}
df['season'] = df['month'].map(season_map)
df['IsSurge'] = (df['deaths_total'] > df['deaths_total'].quantile(0.75)).astype(int)
assert len(df)==442 and int(df['deaths_total'].sum())==4769329


# Tạo SQLite database trong bộ nhớ
conn = sqlite3.connect(':memory:')
df.to_sql('weekly_health', conn, index=False, if_exists='replace')

print(f'Bảng "weekly_health" đã tạo với {len(df)} bản ghi.')
print(f'Các cột: {list(df.columns)}')

# Hàm trợ giúp chạy SQL
def sql(query):
    return pd.read_sql_query(query, conn)
```

```python
# Kiểm tra cấu trúc bảng
sql("PRAGMA table_info(weekly_health)")
```

## 3.2 Tổng số ca tử vong theo từng năm

```python
sql("""
SELECT 
    year AS 'Năm',
    COUNT(*) AS 'Số tuần',
    ROUND(SUM(deaths_total), 0) AS 'Tổng tử vong',
    ROUND(AVG(deaths_total), 0) AS 'TB tử vong/tuần'
FROM weekly_health
GROUP BY year
ORDER BY year
""")
```

## 3.3 Năm nào có nhiệt độ trung bình cao nhất?

```python
sql("""
SELECT 
    year AS 'Năm',
    ROUND(AVG(tasmax_mean_weekly_degC), 2) AS 'TB nhiệt độ (°C)',
    ROUND(MAX(tasmax_max_weekly_degC), 2) AS 'Nhiệt độ max (°C)'
FROM weekly_health
GROUP BY year
ORDER BY AVG(tasmax_mean_weekly_degC) DESC
""")
```

## 3.4 Top 10 tuần có số ca tử vong cao nhất

```python
sql("""
SELECT 
    week_start_mon AS 'Tuần bắt đầu',
    week_end AS 'Tuần kết thúc',
    CAST(deaths_total AS INTEGER) AS 'Số tử vong',
    ROUND(tasmax_mean_weekly_degC, 2) AS 'TB nhiệt độ (°C)',
    season AS 'Mùa',
    year AS 'Năm',
    IsSurge
FROM weekly_health
ORDER BY deaths_total DESC
LIMIT 10
""")
```

## 3.5 Trung bình số ca tử vong theo từng mùa (Xuân, Hạ, Thu, Đông)

```python
sql("""
SELECT 
    season AS 'Mùa',
    COUNT(*) AS 'Số tuần',
    ROUND(AVG(deaths_total), 0) AS 'TB tử vong/tuần',
    CAST(MIN(deaths_total) AS INTEGER) AS 'Min',
    CAST(MAX(deaths_total) AS INTEGER) AS 'Max',
    ROUND(AVG(tasmax_mean_weekly_degC), 2) AS 'TB nhiệt độ (°C)'
FROM weekly_health
GROUP BY season
ORDER BY AVG(deaths_total) DESC
""")
```

## 3.6 Có bao nhiêu tuần có nhiệt độ trung bình > 25°C?

```python
sql("""
SELECT 
    COUNT(*) AS 'Số tuần có TB > 25°C'
FROM weekly_health
WHERE tasmax_mean_weekly_degC > 25
""")
```

```python
# Chi tiết các tuần đó
sql("""
SELECT 
    week_start_mon AS 'Tuần bắt đầu',
    ROUND(tasmax_mean_weekly_degC, 2) AS 'TB nhiệt độ (°C)',
    ROUND(tasmax_max_weekly_degC, 2) AS 'Max nhiệt độ (°C)',
    CAST(deaths_total AS INTEGER) AS 'Số tử vong',
    year AS 'Năm'
FROM weekly_health
WHERE tasmax_mean_weekly_degC > 25
ORDER BY tasmax_mean_weekly_degC DESC
""")
```

## 3.7 Top 5 tuần có nhiệt độ thấp nhất và số tử vong tương ứng

```python
sql("""
SELECT 
    week_start_mon AS 'Tuần bắt đầu',
    ROUND(tasmax_mean_weekly_degC, 2) AS 'TB nhiệt độ (°C)',
    ROUND(tasmax_max_weekly_degC, 2) AS 'Max nhiệt độ (°C)',
    CAST(deaths_total AS INTEGER) AS 'Số tử vong',
    season AS 'Mùa',
    year AS 'Năm',
    IsSurge
FROM weekly_health
ORDER BY tasmax_mean_weekly_degC ASC
LIMIT 5
""")
```

## 3.8 Số lượng tuần được phân loại là IsSurge = 1

```python
sql("""
SELECT 
    IsSurge,
    CASE WHEN IsSurge = 1 THEN 'Surge (Bùng phát)' ELSE 'Normal (Bình thường)' END AS 'Phân loại',
    COUNT(*) AS 'Số tuần',
    ROUND(COUNT(*) * 100.0 / (SELECT COUNT(*) FROM weekly_health), 1) AS 'Tỷ lệ (%)',
    ROUND(AVG(deaths_total), 0) AS 'TB tử vong/tuần',
    ROUND(AVG(tasmax_mean_weekly_degC), 2) AS 'TB nhiệt độ (°C)'
FROM weekly_health
GROUP BY IsSurge
ORDER BY IsSurge
""")
```

```python
# IsSurge theo từng năm
sql("""
SELECT 
    year AS 'Năm',
    SUM(CASE WHEN IsSurge = 1 THEN 1 ELSE 0 END) AS 'Số tuần Surge',
    COUNT(*) AS 'Tổng tuần',
    ROUND(SUM(CASE WHEN IsSurge = 1 THEN 1 ELSE 0 END) * 100.0 / COUNT(*), 1) AS 'Tỷ lệ Surge (%)'
FROM weekly_health
GROUP BY year
ORDER BY year
""")
```

## 3.9 Trung bình nhiệt độ theo từng tháng (2016–2024)

```python
sql("""
SELECT 
    month AS 'Tháng',
    ROUND(AVG(tasmax_mean_weekly_degC), 2) AS 'TB nhiệt độ TB (°C)',
    ROUND(AVG(tasmax_max_weekly_degC), 2) AS 'TB nhiệt độ max (°C)',
    ROUND(MIN(tasmax_mean_weekly_degC), 2) AS 'Thấp nhất (°C)',
    ROUND(MAX(tasmax_mean_weekly_degC), 2) AS 'Cao nhất (°C)',
    ROUND(AVG(deaths_total), 0) AS 'TB tử vong/tuần'
FROM weekly_health
GROUP BY month
ORDER BY month
""")
```

## 3.10 Các tuần có nhiệt độ tăng đột ngột > 5°C so với tuần trước

```python
sql("""
WITH temp_change AS (
    SELECT 
        w1.week_start_mon,
        w1.tasmax_mean_weekly_degC AS temp_current,
        w2.tasmax_mean_weekly_degC AS temp_prev,
        ROUND(w1.tasmax_mean_weekly_degC - w2.tasmax_mean_weekly_degC, 2) AS temp_diff,
        w1.deaths_total,
        w1.season,
        w1.year
    FROM weekly_health w1
    JOIN weekly_health w2 ON date(w1.week_start_sat) = date(w2.week_start_sat, '+7 day')
)
SELECT 
    week_start_mon AS 'Tuần',
    ROUND(temp_prev, 2) AS 'Nhiệt độ tuần trước (°C)',
    ROUND(temp_current, 2) AS 'Nhiệt độ tuần này (°C)',
    temp_diff AS 'Chênh lệch (°C)',
    CAST(deaths_total AS INTEGER) AS 'Số tử vong',
    season AS 'Mùa',
    year AS 'Năm'
FROM temp_change
WHERE temp_diff > 5
ORDER BY temp_diff DESC
""")
```

## 3.11 So sánh tổng số tử vong giữa trước và sau 2020

```python
sql("""
SELECT 
    CASE 
        WHEN year < 2020 THEN 'Trước 2020 (2016–2019)'
        ELSE 'Từ 2020 trở đi (2020–2024)'
    END AS 'Giai đoạn',
    COUNT(*) AS 'Số tuần',
    ROUND(SUM(deaths_total), 0) AS 'Tổng tử vong',
    ROUND(AVG(deaths_total), 0) AS 'TB tử vong/tuần',
    CAST(MAX(deaths_total) AS INTEGER) AS 'Max tử vong/tuần',
    ROUND(AVG(tasmax_mean_weekly_degC), 2) AS 'TB nhiệt độ (°C)'
FROM weekly_health
GROUP BY CASE 
    WHEN year < 2020 THEN 'Trước 2020 (2016–2019)'
    ELSE 'Từ 2020 trở đi (2020–2024)'
END
ORDER BY MIN(year)
""")
```

```python
# Chi tiết hơn: chia 3 giai đoạn (Trước COVID / COVID / Sau COVID)
sql("""
SELECT 
    CASE 
        WHEN year < 2020 THEN '1. Trước COVID (2016–2019)'
        WHEN year BETWEEN 2020 AND 2021 THEN '2. COVID (2020–2021)'
        ELSE '3. Sau COVID (2022–2024)'
    END AS 'Giai đoạn',
    COUNT(*) AS 'Số tuần',
    ROUND(SUM(deaths_total), 0) AS 'Tổng tử vong',
    ROUND(AVG(deaths_total), 0) AS 'TB tử vong/tuần',
    CAST(MAX(deaths_total) AS INTEGER) AS 'Max tử vong/tuần',
    SUM(IsSurge) AS 'Số tuần Surge'
FROM weekly_health
GROUP BY CASE 
    WHEN year < 2020 THEN '1. Trước COVID (2016–2019)'
    WHEN year BETWEEN 2020 AND 2021 THEN '2. COVID (2020–2021)'
    ELSE '3. Sau COVID (2022–2024)'
END
ORDER BY MIN(year)
""")
```

## 3.13 Đối chiếu với bài báo: các con số mô tả dùng để giải thích Hình 1 và Hình 3

Bài báo không lấy một Hình riêng từ Notebook SQL, nhưng các truy vấn ở đây là **kiểm tra độc lập** cho những con số được nêu bên cạnh hình: tuần tử vong cao nhất, tuần nóng nhất, mức trung bình theo mùa và số tuần rất nóng.


```python
from pathlib import Path

def locate_paper_support(relative_path):
    # Accept a self-contained project folder OR the original paper_support/data layout.
    relative_path = Path(relative_path)
    candidates = []
    for root in (Path.cwd(), Path.cwd().parent, Path.cwd().parent.parent):
        candidates.extend([
            root / relative_path.name,
            root / relative_path,
            root / 'paper_support' / relative_path,
            root / 'Datasets' / 'Observed' / relative_path.name,
            root / 'Datasets' / 'Prepared' / relative_path.name,
            root / 'Datasets' / 'Regional_Reference' / relative_path.name,
            root / 'Datasets' / 'Documentation' / relative_path.name,
        ])
    for path in candidates:
        if path.is_file():
            return path
    raise FileNotFoundError(
        f'Missing data file: {relative_path.name}. Put all six notebooks and the five supplied datasets in the same folder. '
        f'Current directory: {Path.cwd()}'
    )

import pandas as pd
import sqlite3
from IPython.display import display

paper_df = pd.read_csv(locate_paper_support('data/uk_weather_health_weekly_2016_2024_TOTAL_FINAL.csv'))
paper_conn = sqlite3.connect(':memory:')
paper_df.to_sql('weekly_health_paper', paper_conn, index=False, if_exists='replace')

queries = {
    'Tuần tử vong cao nhất': '''
        SELECT week_start_mon, CAST(deaths_total AS INTEGER) AS deaths_total
        FROM weekly_health_paper ORDER BY deaths_total DESC LIMIT 1
    ''',
    'Tuần nhiệt độ trung bình cao nhất': '''
        SELECT week_start_mon, ROUND(tasmax_mean_weekly_degC, 2) AS temperature_c
        FROM weekly_health_paper ORDER BY tasmax_mean_weekly_degC DESC LIMIT 1
    ''',
    'Trung bình theo mùa': '''
        SELECT
            CASE
                WHEN CAST(strftime('%m', week_start_mon) AS INTEGER) IN (12,1,2) THEN 'Winter'
                WHEN CAST(strftime('%m', week_start_mon) AS INTEGER) IN (3,4,5) THEN 'Spring'
                WHEN CAST(strftime('%m', week_start_mon) AS INTEGER) IN (6,7,8) THEN 'Summer'
                ELSE 'Autumn'
            END AS season,
            ROUND(AVG(deaths_total), 0) AS avg_deaths,
            ROUND(AVG(tasmax_mean_weekly_degC), 2) AS avg_temperature_c
        FROM weekly_health_paper
        GROUP BY season
        ORDER BY CASE season WHEN 'Winter' THEN 1 WHEN 'Spring' THEN 2 WHEN 'Summer' THEN 3 ELSE 4 END
    ''',
    'Số tuần nhiệt độ trung bình > 25°C': '''
        SELECT COUNT(*) AS hot_weeks
        FROM weekly_health_paper
        WHERE tasmax_mean_weekly_degC > 25
    ''',
}
for title, q in queries.items():
    print('\n' + title)
    display(pd.read_sql_query(q, paper_conn))
paper_conn.close()
```
